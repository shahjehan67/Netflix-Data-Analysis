```python
import pandas as pd
df = pd.read_csv('netflix_titles.csv')
df.head()
```

```python
print("Total Rows and Columns:", df.shape)
print("\nColumn Names:\n", df.columns)

df.info()
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")

plt.figure(figsize=(6, 4))
sns.countplot(x='type', data=df, palette='Set2')

plt.title('Count of Movies vs TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Count')

plt.show()
```

```python
plt.figure(figsize=(12, 6))

df_recent = df[df['release_year'] >= 2000]

sns.countplot(x='release_year', data=df_recent, palette='viridis')

plt.title('Content Release Year Trend on Netflix (2000 onwards)', fontsize=14)
plt.xlabel('Release Year', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.xticks(rotation=90) 

plt.show()
```

```python
plt.figure(figsize=(10, 5))

top_countries = df['country'].value_counts().head(10)

sns.barplot(x=top_countries.values, y=top_countries.index, palette='crest')

plt.title('Top 10 Countries with Most Content on Netflix', fontsize=14)
plt.xlabel('Number of Movies & TV Shows', fontsize=12)
plt.ylabel('Country', fontsize=12)

plt.show()
```
